.intercept_=
The Entrance Fee Example
Imagine you are playing a game where you want to score points.

The game starts you at -2 points (the intercept).

Every time you complete Feature 1, you get +2 points (the coefficient).

Every time you complete Feature 2, you get +2 points (the coefficient).

If you do nothing (Features = 0), your score is -2. You need to successfully complete features just to get your score back up to zero and into positive numbers.
